```python
import pandas as pd

df = pd.read_csv("../data/Road accident.csv")
df.columns
```

```python
df['Accident_Severity'].value_counts()
```

```python
df['Accident_Severity'] = df['Accident_Severity'].replace('Fetal', 'Fatal')
```

```python
total = len(df)

fatal = len(df[df['Accident_Severity'] == 'Fatal'])
serious = len(df[df['Accident_Severity'] == 'Serious'])
slight = len(df[df['Accident_Severity'] == 'Slight'])

print("Total accidents:", total)
print("Fatal rate:", fatal/total)
print("Serious rate:", serious/total)
print("Slight rate:", slight/total)
```

```python
print("Serious from total:", serious/total)
print("Fatal from serious:", fatal/(serious+fatal))
```
